# 2. Principios Físicos

En la clase pasada vimos que la radiación electromagnética emitida por el Sol, interactúa con la atmósfera y la superficie terrestre para luego ser sensada en la plataforma satelital. En esta clase, ampliaremos un poco más estos conceptos.

## Radiación electromagnética

Nos referimos por **radiación electromagnetica** a toda energía que se mueve a la velocidad de la luz en una onda de propagación armónica, con amplitud $A$, longitud de onda $\lambda$ y frecuencia $f$.

Las ondas de mayor longitud tendran menor frecuencia, por lo tanto, menos energía y viceversa.

imagen.png

Las ondas electromagnetica se clasifican en función de su longitud, lo cual resulta de gran utilidad para identificar distintos blancos, ya que interactuan de distinta manera.

En el siguiente gráfico, vemos representadas tres longitudes de onda que corresponden al **espectro visible**, rango al cual nuestros ojos son sensibles a la radiación electromagnética.

image.png

Se puede ver que:

* $\lambda=655\text{ nm}$ corresponde al color **rojo**.
* $\lambda=560\text{ nm}$ corresponde al color **verde**.
* $\lambda=480\text{ nm}$ corresponde al color **azul**.

A valores de longitud de onda inferiores a los $480$ nanómetros se encuentra la **radiación ultravioleta**, luego los rayos X y los rayos gamma de alta energía.

Por encima de los $700$ nanómetros se encuentra la **radiación infrarroja**. A medida que aumenta la longitud de onda, en el orden de los centímetros aparecen las microondas y, en el orden de los metros aparecen las ondas de radio.


image.png

Nuestros ojos sólo sensan el rango visible, pero se pueden fabricar sensores que adquieran información en otras zonas del espectro electromagnético.
Por ejemplo, es posible diferenciar tipos de vegetación o suelos por su comportamiento en el infrarrojo o en las microondas.

Por lo tanto, es fundamental sensar en función del comportamiento espectral de la superficie terrestre.

### Radiancia

> **Radiancia**: flujo de radiación o energía que refleja un área de superficie terrestre en una determinada dirección y ángulo, recibidos durante un tiempo de medición.

De manera coloquial, podemos definirla como el "cono" de flujo de radiación que recibe el sensor proveniente de cierta área.

Cabe destacar que la radiancia depende tanto de las características fisicoquímicas del objeto como de la geometría de observación y la fuente de iluminación.

## Interacción Radiación Electromagnética y Superficie Terrestre

Ahora nos centraremos entre las interacciones que ocurren entre la radiación electromagnetica, la superficie terrestre y la atmósfera. Cuando una onda electromagnética incide sobre una superficie terrestre (suelo, vegetación, agua, etc.), su energía **se reparte en tres componentes**.

image.png

imagen.png

### Reflactancia $(\rho)$

> Proporción de energía que rebota o se **refleja** de la superficie.

* Es lo que detectan principalmente los sensores pasivos como Landsat, Sentinel-2, MODIS, etc.

* Varía con el tipo de superficie:

  *  Alta reflectancia: nieve, salares, suelos secos claros.

  * Baja reflectancia: agua, vegetación densa.

* En vegetación, el infrarrojo cercano (NIR) tiene muy alta reflectancia.

Se define matemáticamente como el cociente entre la radiancia incidente y la radiancia reflejada por el cuerpo,

$$
\rho_\lambda=\frac{L_{p,λ}}{L_{0,λ}}
$$

Para la teledetección óptica, la reflectancia sera la magnitud más importante. Estudiar su comportamiento nos permite hacer inferencias y predicciones sobre la composición y respuesta de los objetos en la superficie terrestre.

### Absorvencia $(α)$

>Proporción de energía que es **absorbida** por la superficie.

* Esta energía puede luego ser:

  *  Reirradiada como calor (infrarrojo térmico).

  * Utilizada en procesos físicos/químicos (ej., fotosíntesis).

* Agua, vegetación húmeda o asfalto absorben mucho en ciertas bandas.

Se define como el cociente entre la radiancia incidente y la radiancia absorvida por el cuerpo,

$$
A_λ=\frac{L_{a,λ}}{L_{0,λ}}
$$

### Transmitancia $(τ)$

> Proporción de energía que **atraviesa** la superficie.



  * En superficies transparentes o semitransparentes (como agua, hielo, hojas), una parte de la energía puede pasar a través.

  * En suelos u objetos opacos, la transmitancia suele ser despreciable $(τ ≈ 0)$.

Matemáticamente, definimos a la transmitancia como el cociente entre la radiancia incidente $L_{t,\lambda}$ y la radiancia transmitida por el cuerpo, $L_{0,\lambda}$:

$$
T_\lambda=\frac{L_{t,\lambda}}{L_{0,\lambda}}
$$

### Firma espectral

El siguiente gráfico, se denomina comúnmente como **firma espectral**, donde se observa la reflectancia para una determinada cobertura homogénea de vegetación, es decir, cuánta radincia refleja del total de la radiancia incidente.

image.png

En el eje $x$ se encuentra la longitud de onda y en el eje $y$ el valor de reflectancia. Por ejemplo, para los 865 nanometros la reflectancia tiene un valor de 0.6 que significa que refleja un 60% de la radiación incidente y el 40% restante puede haber sido absorvido o transmitido para esa longitud de onda.

Si analizamos el comportamiento espectral, vemos un pico de reflectancia entre los 800 y los 1000 nanómetros, que se relaciona con la estructura foliar de la planta.

Por otro lado observamos un valle de absorción cercano a los 660 nanómetros en el rojo y un pico de reflectancia en los 560 nanómetros, este fenómento está relacionado con el contenido de clorofila, que es el pigmento que hace posible la fotosíntesis en las plantas.

Desde los 1500 a los 2200 nanómetros la vegetación muestra dos picos que dependen del contenido de agua foliar.



Entonces, a partir del análisis de una firma espectral, podemos inferir variaciones en el contenido de clorofila, estado foliar y contenido de agua. Asimismo, podremos distinguir tipos de coberturas vegetales comparando sus firmas espectrales. Esto nos da gran ventaja, ya que podemos realizar inferencias a gran escala sobre el estado de la vegetación, analizando el comportamiento espectral.

En el siguiente gráfico, vemos la **firma espectral del agua**:

Su baja reflectancia en todo el espectro electromagnético se debe a su gran capacidad absortiva. En el espectro visible suele ser levemente mayor debido a las partículas disueltas o microorganismo con pigmento fotosintético. Cuanto mayor es la concentración de partículas, mayor es la reflectancia en el espectro visible, decreciendo desde el azul hasta el rojo.

Variables como, el tipo de suelo, la composición, el porcentaje de  materia orgánica o el contenido de agua, son las que modelan el comportamiento de la reflectancia. Por lo tanto, a través de su análisis es posible determinar algunas características fisico-químicas.

## Interpretación Visual de Imágenes

Hablaremos ahora sobre la interpretación visual imágenes satelitales.

Si bien la mayoria de nosotros ya tiene experiencia interpretando imágenes fotográficas en nuestra vida cotidiana, la interpretación de fotografía aérea o imágenes satelitales presentan diferencias sustanciales, como ser:

* Observación de objetos que no nos son familiares.

* La utilización de zonas del espectro en los cuales nuestros ojos no pueden censar.

* La representación de la superficie terrestre a diferentes escalas y resoluciones.

Si bien todos estos parámetros son habituales para un usario con experiencia, para el usuario inicial pueden presentar un gran desafio.



Para la mayoría de las aplicaciones se analizan criterios como,


* **Forma**

* **Patron**

* **Tamaño**

* **Textura**

* **Brillo**

* **Color**

### Forma

Comencemos con la forma.

Nos referiremos al contorno de un objeto individual. Si bien parece algo sencillo determinar formas de objetos desde el espacio puede no serlo para un ojo que no está habituado a esa perspectiva. En la siguiente imágen, vemos algunos ejemplos.

### Tamaño

El tamaño es una de las características más importantes y distintivas. Los parámetros más comunes a medir son el *ancho*, *perímetro* y la *superficie* de un objeto.

El usuario deberá medir de manera habitual los objetos, por lo tanto es imprescindible saber la **escala** y la **resolución espacial**, ya que sino, podran generarse confusiones.

### Textura

La textura refiere al *arreglo de repeticiones de tono o color* en una imágen.

Puede ocurrir que dos coberturas tengan las mismas características espectrales, pero al evaluarlos visualmente, exhiben texturas distintas, permitiendo su diferenciación.

Se habla de **textura suave**, para definir sitios  homogéneos o uniformes y se habal de **textura rugosa**, para identificar zonas gruesas o heterogéneas.

En la siguiente imagen vemos tres ejemplos:

### Brillo

Materiales de la superficie como vegetación, agua o rocas, reflejan diferentes proporciones de energia del espectro electromagnético.
Esta información permite cuantificar la proporción de energía reflejada para cada longitud de onda.

El brillo se observa en una imagen monocromática, es decir, obtenida para una longitud de onda. Las celdas más brillantes representan mayor radiancia reflejada, las más oscuras representan menor radiancia reflejada y las zonas grises son intermedias.

### Color

Las coberturas terrestres reflejan de manera diferente en el espectro electromagnético. Por lo tanto, se pueden inferir algunas características biofísicas a partir de la interpretación visual, relacionandolo con el concepto de firma espectral.

Veamos el siguiente ejemplo con tres firmas espectrales.

image.png

La curva en verde, corresponde a vegetación. La curva naranja, suelo sin cobertura y la azul al agua. Supongamos ahora que tenemos un satélite que puede sensar en 480 nm, longitud correspondiente al azul; en 560 nm, longitud correspondiente al verde; y en 655 nm, longitud correspondiente al rojo.

La información de cada zona censada se almacena en un archivo físico denominado *imágen espectral* o *imagen raster*. Por convención, de ahora en adelante las denominaremos **bandas**.

Entonces, en este caso tendríamos tres bandas, censadas por el satélite.

Los sistemas electrónicos de visualización, como el monitor, utilizan la **teoria aditiva del color** para formar las imágenes, partiendo los colores primarios. Los visualizadores de imágen satelital funcionan de la misma manera, utilizando tres canales: rojo, verde y azul, y mediante su combinación, general el resto de los colores.

image.png

La intensidad de brillo de las bandas que forman parte de la imágen pueden ser mostradas en azul, verde o rojo.

El ejemplo más sencillo se denomina **color real**, en el cual, las bandas correspondientes a 480 nm, longitud del azul; 560 nm longitud del verde; y 655 nm, longitud del rojo, se asignan en el canal azul, verde y rojo respectivamente, en el monitor.

Como resultado se obtiene una imágen tal cual observan nuestros ojos:

image.png

Los distintos tipos de vegetación se visualizan en tonos de verde. Si prestamos atención a la firma espectral de vegetación, vemos que consistentemente, presenta un pico de reflectancia en 560 nm y dos valles de absorción en 480 nm y 655 nm.  Este comportamiento espectral está relacionado con el contenido de clorofila de la hoja.

Los cursos de agua profundos y con baja concentración de partículas en suspensión, se observan en azul oscuro. Aquellos con alta concentración de partículas en suspensión, aparecen en tonos de marrón y celeste, consistente con el pequeño pico de reflectancia que manifiesta la firma espectral.

Los suelos sin cobertura vegetal se muestran en tonos marrones y ocre, relacionado con la longitud de onda más reflectiva del visible.

Cómo se mencionó anteriormente, las plataformas satelitales pueden sensar otras zonas del espectro electromagnético, como el infrarrojo o las microondas, las cuales también pueden ser mostradas en color para analizar visualmente su reflectancia.

En este caso, los distintos tipos de vegetación se visualizan en tonos de rojo y rosado, dado el pico intenso de reflectancia entre los 700 y los 1100 nanómetros.

Las variaciones en esta zona del espectro están relacionadas con la estructura celular de la planta y la densidad del canopeo.
